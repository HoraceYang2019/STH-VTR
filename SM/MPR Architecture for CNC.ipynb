## Objective: One Design, Various Machines, and Identical Parts  

### Background:
    * Various Machines produced by different vendors 
    * Different configurations by machines (accelerometer, current sensor, strain gauge)
    * Multiple benchmarks for evaluating manufacturing capability.

### Architecture Triangle: Physics, Metrology, and Compensation 

### Stages:
* S1. Get the related designs of ISO/Industry benchmark <br> 
* S2. Identify invariant and varying parameters/variables by producing benchmarks <br>
* S3. Evaluate the baseline by the governing equations based on the parameters (spindle speed, feed rate, flute number, tool diameter, ) -> Physics <br>
* S4. Measure the actual variables by the In-suite metrology model with machining learning and AI. -> Metrology <br>
* S5. Compensate the errors between the design and the actual part when applying to different machines. <br>

### Questions:
#### Tool RUL
> * How many workpieces can be finished before replacing a tool? 
> * How long time will it take before replacing the tool? 
> * How about the confidence interval of the previous answer?

### STH availability
> * What is the reliability of the smart tool-holder according to the data collected from calibration and usage?

### Integrate STH and HiNC
Can we only 
> * collect the data for deriving the coefficients of cutting forces from a STH by one run, 
> * find the coefficients using the HiNC, 
> * get the control limits of a run, 
> * and evaluate the cutting forces of the following runs.

### Idea: 
>Step 1. Get the quality and force baselines using the domain features with a built machine learning model. <br> 

>Step 2. Synchronize the force features based on the actual force from STH and the coordinates from the controller. <br> 

>Step 3. Estimate the tool profile by the updated data from STH. <br>

>Step 4. Estimate Quality <br> 

    * Indicate the machining indices (max torque, unbalance, round-out).
    * Predict the surface roughness  
    * Inform an event if out of limits.

>Step 5. Predict tool wear by runs from the wear equation with the initial conditions from the tool profile and affected area.<br>

>Step 6. Simulate cutting force using the predicted tool wear after runs.<br>

>Step 7. Tune the process parameters 

>Step 8. Update limits of the force baseline. <br>
