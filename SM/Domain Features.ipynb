#### 1. Volumetric energy density ($\rho_{ve}$):

\begin{equation}
\rho_{ve} = \frac{p_l}{v_s\left(\pi r_l^2\right)}. \tag{1}
\end{equation}
where<br>
>$p_l$: laser power <br>
$ν_s$: scanning speed <br>
$r_l$: laser beam radius. <br>

#### 2. Richardson number (R):
$$
R = \frac{g{l_{mp}}}{\left(v_g - v_l\right)^2} . \tag{2}
$$
where<br>
>$g$: gravitational acceleration <br>
${l_{mp}}$: length of melt pool <br>
$v_g$: velocity of shielding gas <br>
$v_l$: maximum velocity of the convective flow of the liquid metal <br>



#### 3. Marangoni number (M):
$$
M = -\frac{d\gamma}{dT} \frac{ L \Delta T}{\mu \alpha} . \tag{3}
$$
where<br>
>$\gamma$: surface tension of the alloy <br>
$T$: temperature of the alloy<br>
$μ$: viscosity of the alloy<br>
$α$: thermal diffusivity of the alloy


#### 4. Pool aspect ratio ($ε_{mp}$): 
$$
ε_{mp}= \frac{l_{mp}}{d_{mp}} .\tag{4}
$$
> $l_{mp}$: pool length <br>
$d_{mp}$: pool depth.

#### 5. Surface tension force ($F_{st}$): 
> Surface tension force ($F_{st}$) is calculated as the product of the coefficient of surface tension of an alloy and the circumference of the molten pool on the top surface.
$$
F_{st}=  \gamma \cdot A. \tag{5}
$$
> where <br>
A: circumference of the molten pool 

#### 6. Solidification time of molten pool ($t_s$): 
>Solidification time ($t_s$) is represented by the ratio of pool length to the scanning speed.
