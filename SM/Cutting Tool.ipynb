#### <font color='green'> How to Estimate Cutting Force <font>
The cutting force can be expressed as a function of the cutting area A (which is the product of the chip width, b, and thickness, h), and specific force, K:
    
\begin{equation}
F = K A = K h b. \tag{2}
\end{equation}

For a point on the ($j$th) cutting tooth, differential milling forces corresponding to an infinitesimal element thickness (dz) in the tangential (切線), $dF_t$, 
radial (徑向), $dF_r$, and axial (軸向), $dF_a$, directions can be given as
\begin{equation}
dF_{tj}(\theta,z) = K_t h_j(\theta,z)dz \\
\end{equation}
\begin{equation}
dF_{rj}(\theta,z) = K_r h_j(\theta,z)dz \\
\end{equation}
\begin{equation}
dF_{aj}(\theta,z) = K_a h_j(\theta,z)dz \\
\end{equation}
where $h_j(\theta, z)$ is the instantaneous chip thickness, and $\theta$ is the immersion angle measured from the positive y-axis.

<div>

</div>

Consider $K_{tc}$, $K_{rc}$, and $K_{ac}$ = tangential, radial and axial cutting force coefficients in milling.  <br>
$K_{te}$, $K_{re}$, $K_{ae}$ = tangential, radial and axial edge force coefficients in milling. Then we have

\begin{equation}
dF_{tj}(\theta,z) = [K_{tc} f_t \sin \theta(z) + K_{te}] \cdot dz \\
\end{equation}
\begin{equation}
dF_{rj}(\theta,z) = [K_{rc} f_t \sin \theta(z) + K_{re}] \cdot dz \\
\end{equation}
\begin{equation}
dF_{aj}(\theta,z) = [K_{ac} f_t \sin \theta(z) + K_{ae}] \cdot dz \\
\end{equation}
When in a linear milling process, $h_j(\theta,z) = f_t \sin \theta(z) $ is the immersion angle measured from the positive y-axis, 
chip thickness is determined. $f_t$ is the feed per tooth [mm/tooth].

#### Oblique cutting model for a milling tooth element [1]

The milling force coefficients [1]:
\begin{equation}
K_{tc}, K_{rc}, K_{ac} = \text{functions} (\tau, \alpha_n, i, \phi_n \;\text{or}\; \gamma_l, \beta_n \;\text{or}\; \beta) \;\text{or}, \\
\end{equation} 
\begin{equation}
K_{tc}, K_{rc}, K_{ac} = \text{functions} (\tau, \alpha_n, i, \phi_n \;\text{or}\; \gamma_l, \beta_n \;\text{or}\; \eta_c) \;\text{or}, \\
\end{equation}
\begin{equation}
K_{tc}, K_{rc}, K_{ac} = \text{functions} (\tau, \alpha_n, i, \beta_n \;\text{or}\; \beta,  \eta_c)\\
\end{equation}

where <br>
$\tau$ = $\frac{F}{A}$: shear stress (剪切應力) at the shear plane <br>
$\alpha_n$: normal rake angle (法向傾角) <br>
$i$: helix angle (螺旋角) or angle of obliquity <br>
$\phi$: shear angle (剪切角) in orthogonal cutting <br>
$\phi_n$, $\beta_n$: normal shear and friction angles in oblique cutting, respectively <br>
$\beta$: friction angle at the rake face <br>
$r_l$, $r_t$: chip length and thickness ratios in orthogonal cutting, respectively <br>
$\eta_c$: chip flow angle in the rake face
    
Thus these coefficients can be evaluated if the elemental tool geometry $\alpha_n$, $i$, the work material shear stress $\tau$ and any two
of the three basic cutting quantities {$\phi_n$, $r_l$}, {$\beta$, $\beta_n$}, and $\eta_c$ are known.

- The cutting forces consist of three components including radial force, tangential force, and axial force. 
- The feed (進給, X axis) direction is the direction that is parallel to the tool movement direction, and perpendicular to the tool axis. 
- The normal (正交, Y axis) direction is perpendicular to the tool movement direction and tool axis.
- The axial (軸向, Z axis) direction is parallel to the tool axis and perpendicular to the tool path.

The elemental forces in feed, normal, and axial force are calculated by using the transformation as follows
\begin{equation}
\begin{split} 
   \begin{bmatrix}
           dF_{fj}(\theta,z) \\
           dF_{nj}(\theta,z) \\
           dF_{aj}(\theta,z)
    \end{bmatrix}  =
    \begin{bmatrix}
    -\cos \theta_j(z) & -\sin \theta_j(z) & 0 \\
    \sin \theta_j(z)  & -\cos \theta_j(z) & 0 \\
        0             &      0            & 1
    \end{bmatrix} 
    \begin{bmatrix}
           dF_{tj}(\theta,z) \\
           dF_{rj}(\theta,z) \\
           dF_{aj}(\theta,z) 
    \end{bmatrix}
\end{split} 
\end{equation}

In general, when determining the total cutting force (單刃之切削力), the differential cutting forces are integrated analytically along the in-cut
portion of the flute $j$, as 
\begin{equation}
F_q(\theta_j(z)) = F_q(\theta,z) = \int\limits_{z_{j,1}}^{z_{j,2}} d F_{qj}(\theta, z), q=f,n,a
\end{equation}
where <br>
The values $z_{j,1}$ and $z_{j,2}$ are the lower and upper axial engagement limits of the in-cut portion for the flute number
$j$.

The average cutting forces of flute $j$ per revolution can be expressed as
\begin{equation}
F_{qj}(\theta_j) = \frac{N_f}{2\pi} \int_{\theta_{st}}^{\theta_{ex}+\psi_a} \int_{z_{j,1}}^{z_{j,2}} dF_{qj}(\theta(z)) d\theta, \; q=f,n,a
\end{equation}
where <br>
$\theta_{st}$: The cutter entry angle [deg] <br>
$\theta_{ex}$: The cutter exit angle [deg] <br>
$\psi$, $\psi_a$: The lag angle at axial depth of cut z and z=a, respectively [deg]

Considering the case that has more than one tooth executing the cutting processes simultaneously, the total cutting
forces (每轉一圈之切削力) on the feed, normal, and axial direction can be determined as 
\begin{equation}
F_q(\theta) = \sum\limits_{j=1}^{N_f} F_{qj}(\theta), q=f,n,a
\end{equation}
where $N_f$: the number of flutes on the cutter.

The average forces $\overline{F}_{f}$, $\overline{F}_{n}$, and $\overline{F}_{a}$, can be obtained
experimentally from milling tests.
\begin{equation}
\begin{cases}
\overline{F}_{f} &= -K_{te}S + K_{re}T - \frac{f_t}{4} (-K_{tc}P + K_{rc}Q)\\
\overline{F}_{n} &= -K_{te}T - K_{re}S + \frac{f_t}{4} (-K_{tc}Q + K_{rc}P)\\
\overline{F}_{a} &= -\frac{aN_f}{2\pi} K_{ae}(\theta_{ex}-\theta_{st}) + f_t K_{ac}T \\
\end{cases}
\end{equation}

where <br>
\begin{equation}
\begin{cases}
P &= \frac{aN_f}{2\pi} [\cos 2\theta]_{\theta_{st}}^{\theta_{ex}}\\
Q &= \frac{aN_f}{2\pi} [2\theta - \sin 2\theta]_{\theta_{st}}^{\theta_{ex}}\\
S &= \frac{aN_f}{2\pi} [\sin \theta]_{\theta_{st}}^{\theta_{ex}}\\
T &= \frac{aN_f}{2\pi} [\cos \theta]_{\theta_{st}}^{\theta_{ex}}\\
\end{cases}
\end{equation}

Since slot milling experiments are conducted, the entry and exit angles are $\theta_{st}$ = 0 and $\theta_{ex} = \pi$, respectively. The
parameters P, Q, S, T are constant for all experiments.

The average forces at each feed rate are measured, and edge cutting components ${F}_{qe}$ and ${F}_{qc}$ are estimated by a linear regression
of the data. The average milling forces per tooth period in all direction can be derived as follows.
\begin{equation}
\begin{split}
\begin{cases}
K_{tc} &= 4 \frac{P\overline{F}_{fc} \;+\; Q \overline{F}_{nc}}{P^2+Q^2} \\
K_{rc} &= \frac{PK_{tc}  \;- \; 4 \overline{F}_{fc}}{Q} \\
K_{ac} &= \frac{\overline{F}_{fc}} {T} \\
\end{cases}
\begin{cases}
K_{te} &= \frac{S\overline{F}_{fe}  \;+ \; T \overline{F}_{ne}}{S^2+T^2} \\
K_{re} &= \frac{SK_{te}  \;+ \; \overline{F}_{fe}}{T} \\
K_{ae} &= -\frac{2\pi}{aN_f} \frac{\overline{F}_{ae}} {\theta_{ex}-\theta_{st}} \\
\end{cases}
\end{split}
\end{equation}

The average cutting forces can be expressed by a linear function of feed rate $f_t$ and an offset contributed by the edge force component.
\begin{equation}
\overline{F}_{q} = \overline{F}_{qe} + f_t \overline{F}_{qc}, q=f,n,a
\end{equation}

The cutting torque and power due to the tooth $j$ can be determined as follows
\begin{equation}
T_j(\theta) = \frac{K_t f_t R^2}{\tan \beta}[\cos \theta]_{z_{j1}(\theta)}^{z_{j2}(\theta)} \\
\end{equation}
\begin{equation}
P_j(\theta) = \Omega T_j(\theta)
\end{equation}
where <br>
$R$: cutter radius

The weights or contributions of various factors to the increase in cutting forces as tool wear progresses can vary depending on several conditions such as the material being machined, cutting parameters, and tool material. However, researchers and studies have generally identified the following factors and their approximate impact on cutting forces. These weights are usually specific to a particular machining setup, but here’s a typical breakdown:

> <font color='red'>Increased Friction (30-40%)</font>: As the tool becomes dull, friction increases significantly, making this one of the primary contributors to higher cutting forces.

> <font color='red'>Change in Tool Geometry (20-30%)</font>: Wear leads to changes in tool geometry, like rounding of the cutting edge, which directly impacts the cutting process, leading to a substantial increase in cutting forces.

> <font color='red'>Built-up Edge (BUE) Formation (10-20%)</font>: BUE adds to the resistance during machining but is typically less impactful than friction or tool geometry changes, though it can still cause significant variations in force.

> <font color='red'>Increase in Contact Area (10-15%)</font>: As the wear progresses, more of the tool comes into contact with the material, which increases the cutting force but is less influential than friction or geometry changes.

> <font color='red'>Thermal Effects (5-10%)</font>: The increase in temperature due to wear can affect tool performance and material characteristics, but the direct influence on cutting force is typically lower compared to mechanical factors like friction or geometry.

> <font color='red'>Chip Formation Changes (5-10%)</font>: While changes in chip formation can cause irregular cutting and contribute to force increase, this factor has a relatively smaller impact.

> <font color='red'>Surface Finish Degradation (5-10%)</font>: Poor surface finish creates irregularities that contribute to higher forces, though it generally has a lesser effect compared to the major factors like friction and tool geometry changes.

These weights are approximate and would vary in real-world scenarios depending on specific machining conditions. Advanced studies using machine learning or empirical testing can provide more precise weights for a particular machining setup.

Here are the formulas related to the top 3 factors influencing cutting forces during tool wear: increased friction, change in tool geometry, and built-up edge (BUE) formation.

* Increased Friction <br>
As the tool wears, friction between the tool and the workpiece increases, which contributes to the rise in cutting forces. 
The friction force can be calculated using the coefficient of friction (μ) and the normal force (FN) acting on the tool's contact area:<br>
\begin{equation}
F_f = μ⋅F_N
\end{equation}

Where:<br>
$F_f$ is the frictional force. <br>
μ is the coefficient of friction between the tool and the workpiece.<br>
$F_N$ is the normal force, which depends on the cutting depth and feed rate.<br>

As the coefficient of friction (μ) increases with wear, the overall cutting force $F_c$ also increases:<br>
\begin{equation}
F_c = 𝐹_f + 𝐹_𝑐^{ideal}. 
\end{equation}
Where  $F_c^{ideal}$ is the cutting force without wear.

In theory, based on the formula from Merchant’s Circle (𝜇= $\frac{sin(𝜙)}{cos⁡(𝜙−𝛼)}$), the coefficient of friction (𝜇) would appear to be fixed if the shear angle (ϕ) and rake angle (α) are fixed. 
However, this assumption doesn't fully account for the effects of tool wear and other dynamic factors during machining.

Here’s why μ may still vary with tool wear:

* 1. Tool Wear Alters the Effective Rake Angle <br>
Wear on the tool (especially flank and crater wear) changes the geometry of the cutting edge, which in turn can alter the effective rake angle (α). <u> A dulled or worn tool typically reduces the effective rake angle </u>, even if the initial design rake angle remains fixed.
As the rake angle decreases due to wear, the cutting edge becomes less sharp, which increases resistance during cutting and thereby increases the friction between the tool and the workpiece. This change in geometry effectively increases μ.

* 2. Friction Changes at the Tool-Workpiece Interface<br>
Tool wear increases friction at the tool-chip and tool-workpiece interfaces. <u>As the tool becomes worn, the surface finish of the tool degrades, increasing the coefficient of friction due to rougher contact between the tool and the material.</u>
Wear also leads to heat generation, which can soften the material being cut, change chip formation, and further increase the frictional forces, even though the shear and rake angles remain theoretically fixed.

* 3. Built-Up Edge (BUE) Formation <br>
BUE often forms more readily on worn tools. <u>This temporary accumulation of workpiece material on the cutting edge increases the friction and resistance to cutting</u>, causing a variation in μ. Even though the rake angle and shear angle are fixed, BUE adds an irregular layer that changes the dynamics of the cutting forces.

* 4. Surface Finish and Chip Flow Changes <br>
As the tool wears, the chip formation process becomes less efficient, leading to more complex interactions between the tool and the workpiece material. <u>The shear angle ϕ may not change dramatically, but the nature of chip formation (irregular, thicker, or poorer chip evacuation) adds to frictional forces</u>, impacting μ.
